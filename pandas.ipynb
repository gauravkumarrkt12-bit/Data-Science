## 6. Data Cleaning

```python
# Clean discounted price
df["discounted_price"] = (
    df["discounted_price"]
    .astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
)
df["discounted_price"] = pd.to_numeric(df["discounted_price"], errors="coerce")

# Clean actual price
df["actual_price"] = (
    df["actual_price"]
    .astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
)
df["actual_price"] = pd.to_numeric(df["actual_price"], errors="coerce")

# Clean discount percentage
df["discount_percentage"] = (
    df["discount_percentage"]
    .astype(str)
    .str.replace("%", "", regex=False)
)
df["discount_percentage"] = pd.to_numeric(
    df["discount_percentage"], errors="coerce"
)
 
# Convert rating to numeric
df["rating"] = pd.to_numeric(df["rating"], errors="coerce")

# Clean rating count
df["rating_count"] = (
    df["rating_count"]
    .astype(str)
    .str.replace(",", "", regex=False)
)
df["rating_count"] = pd.to_numeric(df["rating_count"], errors="coerce")

print("Data types after cleaning:")
display(df.dtypes)
```

## 7. Feature Engineering

```python
# Savings
df["savings"] = df["actual_price"] - df["discounted_price"]

# Review length in characters
df["review_length"] = df["review_content"].str.len()

# Review word count
df["word_count"] = df["review_content"].str.split().str.len()

# Main category
df["main_category"] = df["category"].str.split("|").str[0] 

# Approximate brand from first word of product name 
df["brand"] = df["product_name"].str.split().str[0]


# basic idea to check unique value and perform operation using dictionary 



display(df.head()) 
```

## 8. Price Analysis

```python
print("Minimum discounted price:", df["discounted_price"].min())
print("Maximum discounted price:", df["discounted_price"].max())
print("Average discounted price:", df["discounted_price"].mean())
print("Median discounted price:", df["discounted_price"].median())

print("\nMinimum actual price:", df["actual_price"].min())
print("Maximum actual price:", df["actual_price"].max())
print("Average actual price:", df["actual_price"].mean())
print("Median actual price:", df["actual_price"].median())

print("\n10 most expensive products:")
display(df.nlargest(10, "discounted_price")[
    ["product_id", "product_name","brand",  "discounted_price"]
])

print("\n10 cheapest products:")
display(df.nsmallest(10, "discounted_price")[
    ["product_id", "product_name","brand", "discounted_price"]
])


df[df["discounted_price"] == 54990]
```

## 9. Discount Analysis

```python
print("Minimum discount:", df["discount_percentage"].min())
print("Maximum discount:", df["discount_percentage"].max())
print("Average discount:", df["discount_percentage"].mean())

print("\nProducts with no discount:")
display(df[df["discount_percentage"] == 0].nunique()) 

print("\nProducts with more than 50% discount:")
display(df[df["discount_percentage"] > 50].head(5))

print("\nTop 10 most discounted products:")
display(df.nlargest(10, "discount_percentage")[
    [
        "product_id", "product_name",
        "actual_price", "discounted_price",
        "discount_percentage"
    ]
])
 
```

## 10. Discount Groups

```python
df["discount_group"] = pd.cut(
    df["discount_percentage"],
    bins=[-1, 10, 30, 50, 70, 100],
    labels=[
        "0-10%",
        "11-30%",
        "31-50%",
        "51-70%",
        "71-100%"
    ]
)


print("Number of products in each discount group:")
display(df["discount_group"].value_counts().sort_index())

print("\nDiscount group analysis:")
display(
    df.groupby("discount_group", observed=True).agg(
        number_of_products=("product_id", "count"),
        average_price=("discounted_price", "mean"),
        average_rating=("rating", "mean"),
        average_rating_count=("rating_count", "mean")
    )
)
```

## 11. Savings Analysis

```python
print("Average savings:", df["savings"].mean())
print("Minimum savings:", df["savings"].min())
print("Maximum savings:", df["savings"].max())

print("\nTop 10 products by savings:")
display(df.nlargest(10, "savings")[
    [
        "product_name",
        "actual_price",
        "discounted_price",
        "discount_percentage",
        "savings"
    ]
])
```

## 12. Rating Analysis

```python
print("Minimum rating:", df["rating"].min())
print("Maximum rating:", df["rating"].max())
print("Average rating:", df["rating"].mean())
print("Median rating:", df["rating"].median())

print("\nRating distribution:")
display(df["rating"].value_counts().sort_index())

print("\nRating percentage distribution:")
display((df["rating"].value_counts(normalize=True).sort_index() * 100)) 

print("\nProducts with rating 5:")
display(df[df["rating"] == 5][
    ["product_name", "rating", "rating_count"]
])

print("\nProducts with rating below 3:")
display(df[df["rating"] < 3][
    ["product_name", "rating", "rating_count"]
])

```

# 17. POINT PLOT  — Category means with confidence intervals
# ════════════════════════════════════════════════════════════════════

```python
df["price_range"] = pd.cut(
    df["actual_price"],
    bins=[0, 500, 1000, 2000, 5000, float("inf")],
    labels=["0-500", "500-1000", "1000-2000", "2000-5000", "5000+"]
)

plt.figure(figsize=(10, 6))

sns.pointplot(
    data=df,
    x="price_range",
    y="rating",
    hue="main_category",
    dodge=True,
    capsize=0.1,
    markers="o"
)

plt.title("Mean Rating by Actual Price Range & Main Category")
plt.xlabel("Actual Price Range")
plt.ylabel("Mean Rating")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
