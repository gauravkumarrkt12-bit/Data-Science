# ============================================================
# 8. DATA CLEANING
# ============================================================

```python
clean_df = df.copy()
```


# -----------------------------
# Remove duplicates
# -----------------------------

```python

clean_df = clean_df.drop_duplicates()

print("Rows after removing duplicates:", len(clean_df))
```

# -----------------------------
# Convert dates
# -----------------------------

```python
clean_df["Admission_Date"] = pd.to_datetime(
    clean_df["Admission_Date"],
    errors="coerce"
)

clean_df["Discharge_Date"] = pd.to_datetime(
    clean_df["Discharge_Date"],
    errors="coerce"
)
```


# -----------------------------
# Fill categorical missing values
# -----------------------------

```python

categorical_columns = clean_df.select_dtypes(
    include="object"
).columns

for column in categorical_columns:

    if clean_df[column].isnull().sum() > 0:

        clean_df[column] = clean_df[column].fillna(
            clean_df[column].mode()[0]
        )
```


# -----------------------------
# Fill numerical missing values
# -----------------------------

```python

numerical_columns = clean_df.select_dtypes(
    include=np.number
).columns

for column in numerical_columns:

    if clean_df[column].isnull().sum() > 0:

        clean_df[column] = clean_df[column].fillna(
            clean_df[column].median()
        )


print(
    "Remaining missing values:",
    clean_df.isnull().sum().sum()
)
```

# ============================================================
# 9. CREATE NEW COLUMNS
# ============================================================


```python
# Admission year

clean_df["Admission_Year"] = (
    clean_df["Admission_Date"].dt.year
)


# Admission month

clean_df["Admission_Month"] = (
    clean_df["Admission_Date"].dt.month_name()
)


# Admission month number

clean_df["Admission_Month_Number"] = (
    clean_df["Admission_Date"].dt.month
)


# Treatment cost in lakhs

clean_df["Treatment_Cost_Lakh"] = (
    clean_df["Treatment_Cost"] / 100000
)


print("\n15. NEW COLUMNS CREATED")
print("-" * 70)

print(
    clean_df[
        [
            "Admission_Year",
            "Admission_Month",
            "Treatment_Cost_Lakh"
        ]
    ].head()
)
```


# ============================================================
# 12. OUTLIER ANALYSIS USING IQR
# ============================================================

```python



outlier_columns = [
    "Age",
    "Stay_Days",
    "Treatment_Cost",
    "Patient_Satisfaction"
]

for column in outlier_columns:

    Q1 = clean_df[column].quantile(0.25)

    Q3 = clean_df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR

    upper_limit = Q3 + 1.5 * IQR

    outliers = clean_df[
        (clean_df[column] < lower_limit) |
        (clean_df[column] > upper_limit)
    ]

    print(
        f"\n{column}"
    )

    print(
        "Q1:",
        round(Q1, 2)
    )

    print(
        "Q3:",
        round(Q3, 2)
    )

    print(
        "IQR:",
        round(IQR, 2)
    )

    print(
        "Lower Limit:",
        round(lower_limit, 2)
    )

    print(
        "Upper Limit:",
        round(upper_limit, 2)
    )

    print(
        "Number of Outliers:",
        len(outliers)
    )
```

```python


total_patients = clean_df["Patient_ID"].nunique()

total_treatment_cost = (
    clean_df["Treatment_Cost"].sum()
)

average_treatment_cost = (
    clean_df["Treatment_Cost"].mean()
)

average_age = (
    clean_df["Age"].mean()
)

average_stay = (
    clean_df["Stay_Days"].mean()
)

average_satisfaction = (
    clean_df["Patient_Satisfaction"].mean()
)

discharge_rate = (
    clean_df["Hospital_Status"]
    .eq("Discharged")
    .mean()
    * 100
)

emergency_rate = (
    clean_df["Admission_Type"]
    .eq("Emergency")
    .mean()
    * 100
)
```

# ============================================================
# 18. AUTOMATIC INSIGHTS
# ============================================================

```python
department_analysis = clean_df.groupby("Department").agg(
    Patients=("Patient_ID", "nunique"),
    Total_Treatment_Cost=("Treatment_Cost", "sum")
)

city_analysis = clean_df.groupby("City").agg(
    Patients=("Patient_ID", "nunique"),
    Total_Treatment_Cost=("Treatment_Cost", "sum")
)

doctor_analysis = clean_df.groupby("Doctor").agg(
    Patients=("Patient_ID", "nunique")
)

department_satisfaction = (
    clean_df.groupby("Department")["Patient_Satisfaction"]
    .mean()
    .sort_values(ascending=False)
)

highest_cost_department = (
    department_analysis["Total_Treatment_Cost"]
    .idxmax()
)

highest_cost_city = (
    city_analysis["Total_Treatment_Cost"]
    .idxmax()
)

most_patients_doctor = (
    doctor_analysis["Patients"]
    .idxmax()
)

highest_satisfaction_department = (
    department_satisfaction.idxmax()
)


print("\n" + "=" * 70)
print("KEY EDA INSIGHTS")
print("=" * 70)

print(
    "1. Highest treatment-cost department:",
    highest_cost_department
)

print(
    "2. Highest treatment-cost city:",
    highest_cost_city
)

print(
    "3. Doctor handling most patients:",
    most_patients_doctor
)

print(
    "4. Highest satisfaction department:",
    highest_satisfaction_department
)

print(
    f"5. Average patient age: {average_age:.2f} years"
)

print(
    f"6. Average hospital stay: {average_stay:.2f} days"
)

print(
    f"7. Average treatment cost: ₹{average_treatment_cost:,.2f}"
)

print(
    f"8. Average patient satisfaction: "
    f"{average_satisfaction:.2f}/5"
)
```
