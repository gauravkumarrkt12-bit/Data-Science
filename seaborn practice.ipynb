```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python


# Tips dataset
df = sns.load_dataset("tips")

# Display DataFrame
display(df)
```

```python
tips = sns.load_dataset("tips")
flights = sns.load_dataset("flights")
titanic = sns.load_dataset("titanic")
iris = sns.load_dataset("iris")
penguins = sns.load_dataset("penguins")

display(tips)
```

```python
sns.set_theme(style="whitegrid")   
```

# ════════════════════════════════════════════════════════════════════
# 1. SCATTER PLOT  — Relationship between two continuous variables
# ════════════════════════════════════════════════════════════════════

```python
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df, x="total_bill", y="tip",
                hue="sex", size="size", palette="Set1")
plt.title("Scatter Plot — Total Bill vs Tip")
plt.tight_layout()
plt.show()
```

# ════════════════════════════════════════════════════════════════════
# 2. LINE PLOT  — Trend over time / ordered variable
# ════════════════════════════════════════════════════════════════════

```python
display(flights)
```

```python

plt.figure(figsize=(10, 5))
sns.lineplot(data=flights, x="year", y="passengers",
             hue="month", palette="tab20", legend=True)
plt.title("Line Plot — Passengers per Year by Month")
plt.tight_layout()
plt.show()
```

# ════════════════════════════════════════════════════════════════════
# 3. BAR PLOT  — Compare category means
# ════════════════════════════════════════════════════════════════════

```python
plt.Figure(figsize=(10,7))
sns.barplot(data= titanic , x="class" , y= "survived" , hue="sex" , palette="muted" , legend=True)
plt.title("bar_chart survived by clases" )
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.barplot(data=tips, x="day", y="total_bill",
            hue="sex", palette="muted", ci="sd")
plt.title("Bar Plot — Average Total Bill by Day & Sex")
plt.tight_layout()
plt.show()
```


# ════════════════════════════════════════════════════════════════════
# 4. COUNT PLOT  — Frequency of categories (like bar chart for counts)
# ════════════════════════════════════════════════════════════════════

```python
plt.figure(figsize=(7, 5))
sns.countplot(data=titanic, x="class", hue="survived", palette="Set2")
plt.title("Count Plot — Titanic Survival by Class")
plt.tight_layout()
plt.show()
```

# ════════════════════════════════════════════════════════════════════
# 5. BOX PLOT  — Distribution + outliers per category
# ════════════════════════════════════════════════════════════════════

```python

plt.figure(figsize=(8, 5))
sns.boxplot(data=tips, x="day", y="total_bill",
            hue="smoker", palette="coolwarm")
plt.title("Box Plot — Total Bill Distribution by Day & Smoker")
plt.tight_layout() 
plt.show()
```

# ════════════════════════════════════════════════════════════════════
# 6. VIOLIN PLOT  — Box plot + KDE (smooth distribution shape)
# ════════════════════════════════════════════════════════════════════

```python

plt.figure(figsize=(8, 5))
sns.violinplot(data=tips, x="day", y="total_bill",
               hue="sex", split=True, palette="muted")
plt.title("Violin Plot — Total Bill by Day & Sex")
plt.tight_layout()
plt.show()
```

# ════════════════════════════════════════════════════════════════════
# 7. HISTOGRAM  — Distribution of a single numeric variable
# ════════════════════════════════════════════════════════════════════

```python

plt.figure(figsize=(7, 5))
sns.histplot(data=tips, x="total_bill", bins=5,
             kde=True, color="steelblue")
plt.title("Histogram — Total Bill Distribution")
plt.tight_layout()
plt.show()
```

# ════════════════════════════════════════════════════════════════════
# 8. KDE PLOT  — Smooth probability density estimate
# ════════════════════════════════════════════════════════════════════

```python

plt.figure(figsize=(7, 5))
sns.kdeplot(data=tips, x="total_bill", hue="sex",
            fill=True, alpha=0.5, palette="Set1")
plt.title("KDE Plot — Total Bill Density by Sex")
plt.tight_layout()
plt.show()
```

# ════════════════════════════════════════════════════════════════════
# 9. ECDF PLOT  — Empirical Cumulative Distribution Function
# ════════════════════════════════════════════════════════════════════

```python


plt.figure(figsize=(7, 5))
sns.ecdfplot(data=tips, x="total_bill", hue="day")
plt.title("ECDF Plot — Cumulative Distribution of Total Bill")
plt.tight_layout()
plt.show()
```
