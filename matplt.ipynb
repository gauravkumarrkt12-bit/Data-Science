```python
# #
# #
# Sample car sales company data
data = {
    'Car_Model': ['Sedan', 'SUV', 'Hatchback', 'Coupe', 'Convertible', 'Truck', 'Wagon', 'Van', 'Crossover', 'Hybrid'],
    'Units_Sold': [1500, 2500, 2000, 800, 400, 1200, 900, 700, 1600, 1000],
    'Average_Price': [25000, 35000, 20000, 40000, 45000, 30000, 28000, 27000, 32000, 30000],
    'Customer_Satisfaction': [8.5, 9.2, 8.8, 8.0, 7.8, 8.6, 8.3, 8.0, 8.9, 9.5],
    'Profit_Margin': [15, 20, 18, 12, 10, 14, 16, 13, 19, 25],
    'Yearly_Growth': [5, 7, 6, 4, 3, 5, 4, 3, 6, 8]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

## bar chart

```python
# plot 1 : bar chart of  u its sold per car model

plt.figure(figsize=(10,6))
bars = plt.bar(df['Car_Model'],df['Units_Sold'],color = 'blue',edgecolor = 'black', linewidth = 1.2 ,label = 'car_model')

plt.title('unit sold per car model',fontsize = 16 , fontweight = 'bold'  )
plt.gca().set_axisbelow(True)
plt.grid(axis='y' , linestyle= '--',alpha=0.5)
plt.xlabel('car_model', fontsize= 14 , fontweight = 'bold')
plt.ylabel('unit_sold', fontsize=14,fontweight='extra bold')
plt.xticks(rotation=90 ,fontsize= 14)
plt.legend(title= 'cars')
plt.tight_layout()
plt.show()
```

line chart

```python
# plot 2 : line chart of average price for each car model


# Plot 2: line chart of avg price for each  car model
plt.figure(figsize=(10, 6))
plt.plot(df['Car_Model'], df['Average_Price'], marker= '*',color="green", linewidth=1.2, label="Average Price")
plt.title('Average Price per Car Model', fontsize=16, fontweight='bold')
plt.gca().set_axisbelow(True)
plt.grid(axis='y', linestyle='--', alpha=1)
plt.xlabel('Car Model', fontsize=14, fontweight='bold')
plt.ylabel('Average Price', fontsize=14, fontweight='bold')
plt.xticks(rotation=90, fontsize=14)
plt.legend(title='cars')
plt.tight_layout()
plt.show()
```

## scatter chart

```python
# Plot 4: Scatter plot of customer satisfaction vs profit margin
plt.figure(figsize=(10, 6))
plt.scatter(df['Customer_Satisfaction'], df['Profit_Margin'], color='brown')
plt.title('Customer Satisfaction vs Profit Margin')
plt.xlabel('Customer Satisfaction (Rating)')
plt.ylabel('Profit Margin (%)')
plt.grid(True)
plt.show()
```

##  horizontal bar chart

```python
plt.figure(figsize=(10,6))
Hbars = plt.barh(df['Car_Model'],df['Units_Sold'],color = 'blue',edgecolor = 'black', linewidth = 1.2 ,label = 'car_model')

plt.title('unit sold per car model',fontsize = 16 , fontweight = 'bold'  )
plt.gca().set_axisbelow(True)
plt.grid(axis='y' , linestyle= '--',alpha=0.5)
plt.xlabel('car_model', fontsize= 14 , fontweight = 'bold')
plt.ylabel('unit_sold', fontsize=14,fontweight='extra bold')
plt.xticks(rotation=90 ,fontsize= 14)
plt.legend(title= 'cars')
plt.tight_layout()
plt.gca().set_facecolor('lightblue')
plt.gcf().set_facecolor('lightblue')
plt.show()
```

## histogram chart

```python
 # Plot 6: Histogram of yearly growth percentages

plt.figure(figsize=(10, 6))
plt.hist(df['Yearly_Growth'], bins=2, color='blue', edgecolor='black')
plt.title('Distribution of Yearly Growth')
plt.xlabel('Yearly Growth (%)')
plt.ylabel('Frequency')
plt.gca().set_facecolor('lightblue')
plt.gcf().set_facecolor('lightblue')
plt.show()
```

## Gruop chart

```python


x = np.arange(len(data['Car_Model']))
width = 0.35

plt.figure(figsize=(12, 6))

plt.bar(x - width/2, data['Units_Sold'], width, label='Units Sold')
plt.bar(x + width/2, data['Profit_Margin'], width, label='Profit Margin')

plt.xticks(x, data['Car_Model'], rotation=45)
plt.title('Units Sold vs Profit Margin')
plt.xlabel('Car Model')
plt.ylabel('Value')
plt.legend()

plt.tight_layout()
plt.show()
```

## multiple line comparison

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df['Car_Model'],
    df['Customer_Satisfaction'],
    marker='o',
    label='Customer Satisfaction'
)

plt.plot(
    df['Car_Model'],
    df['Profit_Margin'],
    marker='o',
    label='Profit Margin'
)

plt.plot(
    df['Car_Model'],
    df['Yearly_Growth'],
    marker='o',
    label='Yearly Growth'
)



plt.title('Car Model Performance Comparison' , fontweight='bold')
plt.xlabel('Car Model')
plt.ylabel('Value')
plt.xticks(rotation=45)
plt.legend()

plt.tight_layout()
plt.show()
```

## stacked bar chart !

```python
plt.figure(figsize=(12, 6))



plt.bar(
    df['Car_Model'],
    df['Profit_Margin'],
    label='Profit Margin' , color='brown'
)

plt.bar(
    df['Car_Model'],
    df['Yearly_Growth'],
    label='Yearly Growth' , color='pink'
)

plt.title('Car Model Performance Comparison')
plt.xlabel('Car Model')
plt.ylabel('Value')
plt.xticks(rotation=45)
plt.legend()

plt.tight_layout()
plt.show()
```
